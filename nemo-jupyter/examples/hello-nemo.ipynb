# Nemo in Jupyter

This notebook uses the **Nemo** kernel. Cells **accumulate into one Nemo program**: facts and rules from earlier cells stay active in later cells, like a normal notebook. Every run re-reasons the whole program and prints the contents of each `@export`-ed predicate.

Run the cells in order.

## 1. Facts

Start with facts only. They are added to the program, but nothing is exported yet, so there is no result output.

```nemo
parent(ada, bob) .
parent(bob, cyd) .
parent(cyd, dan) .
```

## 2. Rules

Now add rules that use the facts from the previous cell, and export the result. The summary line shows the cell count of the accumulated program.

```nemo
ancestor(?x, ?y) :- parent(?x, ?y) .
ancestor(?x, ?y) :- parent(?x, ?z), ancestor(?z, ?y) .

@export ancestor :- csv {}.
```

## 3. Debugging with traces

Kernel commands start with `!` (Nemo reserves `%` for comments). `!trace <fact>` shows how a fact was derived in the accumulated program:

```nemo
!trace ancestor(ada, dan)
```

## 4. More facts

Add another fact in a new cell. The accumulated program re-reasons and the new ancestor appears.

```nemo
parent(dan, eve) .
```

## 5. The accumulated program

`!program` shows everything accumulated so far. Useful commands: `!help`, `!reset` (clear the program), `!standalone` (run one cell in isolation), `!predicates`, `!load <file>`, `!trace <fact>`.

```nemo
!program
```
